```r
# ---STRUCTURAL ELEMENTS IN MAIN MODEL---

source_between_markers <- function(file_path, start_marker, end_marker) {
  
        
# Read all lines from the file
  lines <- readLines(file_path, warn = FALSE)

  # Locate the lines containing the start and end markers
  start <- grep(start_marker, lines)
  end <- grep(end_marker, lines)

  if (length(start) == 0 || length(end) == 0) {
    stop("Markers not found in the file.")
  }
  if (start >= end) {
    stop("Start marker appears after end marker.")
  }

  # Extract lines between the markers (excluding the marker lines themselves)
  code_lines <- lines[(start + 1):(end - 1)]

  # Evaluate the code in the global environment
  eval(parse(text = code_lines), envir = .GlobalEnv)
}

source_between_markers(
  file_path = here::here("notebooks", "r-nb", "A-prepSteps", "2-structure.r"),
  start_marker = "## BEGIN Structure",
  end_marker = "## END Structure"
)
```

```r
# ------------ ------------ ------------ ------------ ------------ 
# Functions 

parse_vensim_data <- function(file_path, dim_names) {
  txt <- readLines(file_path, warn = FALSE)
  txt <- gsub("\\\\$", "", txt)
  txt <- paste(txt, collapse = " ")
  
  # ------------------------------------------------
  # -- TABBED ARRAY case ---------------------------
  # ------------------------------------------------
  if (grepl("TABBED ARRAY", txt)) {

    # Find templates in the global environment
    template_names <- grep("^template_", ls(envir = .GlobalEnv), value = TRUE)
    templates <- lapply(template_names, function(nm) get(nm, envir = .GlobalEnv))
    names(templates) <- template_names 
    # Finding template
    matching_templates <- Filter(function(tmpl) {
      is.array(tmpl) &&
      length(dimnames(tmpl)) == length(dim_names) &&
      setequal(names(dimnames(tmpl)), dim_names)
    }, templates)

    if (length(matching_templates) == 0) {
      stop("❌ No matching template found with dimensions: ", paste(dim_names, collapse = ", "))
    }
    if (length(matching_templates) > 1) {
      stop("❌ Multiple matching templates found. Please ensure uniqueness.")
    }
    template <- matching_templates[[1]]

    desc <- gsub("\\t", "", txt)
    # Extraire tout ce qui vient après le premier ~
    desc <- str_replace(desc, ".*?~", "")
    # Si desc est vide (ou contient juste des espaces), mettre NA
    if (str_trim(desc) == "") {
      desc <- NA
    }
    
    # Si on a trouvé la partie TABBED ARRAY, remplacer les tabulations par des virgules
    array_part <- str_match(txt, "TABBED ARRAY\\(\\s*(.*?)\\s*\\)")
    if (!is.na(array_part[1, 2])) {
      tabbed_values <- array_part[1, 2]
      tabbed_values <- gsub("\\t", ",", tabbed_values)
        tabbed_values <- gsub(" ", "", tabbed_values)       # supprime tous les espaces
        tabbed_values <- gsub(",+", ",", tabbed_values)    # remplace 1 ou plusieurs virgules consécutives par une virgule
      #txt <- gsub(array_part[1, 2], tabbed_values, txt)  # Remplacer la partie du texte original
    }
    
    # Supprimer tous les espaces sauf entre les tabulations
    #txt <- gsub("\\s+", "", txt)
    
    # Extraire les dimensions entre crochets
    dim_match <- str_match(txt, "\\[([^\\]]+)\\]")
    if (is.na(dim_match[1, 1])) next  # Si aucune dimension n'est trouvée, passer à l'itération suivante
    dims <- unlist(strsplit(dim_match[1, 2], ","))  # Dimensions entre crochets
    if (length(dims) != 2) {
      stop("❌ The TABBED ARRAY() appears to have more than 2 dimensions. Check the .txt file for corrupted parts.")
    }
    
    # Afficher les dimensions et demander à l'utilisateur de faire des choix
    cat(
      "⚠️ The following dimensions were found: ", dims, "\n",
      "Please match each dimension to an index from 'dim_names'.\n\n",
      "Available options for dimension 1 (", dims[1], "):\n"
    )
    
    # Afficher toutes les options disponibles pour dimension 1
    for (i in seq_along(dim_names)) {
      cat(i, ": ", dim_names[i], "\n")
    }
    flush.console()  # Un seul flush à la fin pour forcer l'affichage
    modality_1_index <- as.integer(readline(paste("❓ Choose the index for dimension 1 (", dims[1], "): ")))

    cat("\nAvailable options for dimension 1 (", dims[1], "):\n")
    for (i in seq_along(dim_names)) {
      cat(i, ": ", dim_names[i], "\n")
    }
    flush.console()  # Assurer l'affichage immédiat
    modality_2_index <- as.integer(readline(paste("❓ Choose the index for dimension 2 (", dims[2], "): ")))

    # Vérifier si l'utilisateur a choisi un index valide
    while (!(modality_1_index %in% seq_along(dim_names))) {
      modality_1_index <- as.integer(readline(paste("❌ Invalid index. Please choose a valid index for dimension 1 (", dims[1], "): ")))
    }
    while (!(modality_2_index %in% seq_along(dim_names))) {
      modality_2_index <- as.integer(readline(paste("❌ Invalid index. Please choose a valid index for dimension 2 (", dims[2], "): ")))
    }

    # Récupérer le nom des dimensions en utilisant l'index choisi
    modality_1 <- dim_names[modality_1_index]
    modality_2 <- dim_names[modality_2_index]

    # Affichage des choix
    cat("✅ You selected the following modalities:\n")
    cat("Dimension 1 (", dims[1], "): ", modality_1, "\n")
    cat("Dimension 2 (", dims[2], "): ", modality_2, "\n")

    # Transform dimName in modName
    modality_name_1 <- dimnames(template)[[which(names(dimnames(template)) == modality_1)[1]]]
    modality_name_2 <- dimnames(template)[[which(names(dimnames(template)) == modality_2)[1]]]
    
    # Crée une matrice vide avec les dimensions
    temp_array <- array(NA, 
                        dim = c(length(modality_name_1), length(modality_name_2)),
                        dimnames = list(modality_1 = modality_name_1, modality_2 = modality_name_2)
    )
    
    # Extraction des valeurs entre TABBED ARRAY( 
    array_values <- str_match(tabbed_values, "([^)]+)")  

    if(dim_names[1] == dim_names[2]){

        value_list <- as.numeric(unlist(strsplit(array_values[1, 2], ",")))
        # Remplir la matrice avec les valeurs extraites
        temp_array[] <- value_list  
        dname <- names(dimnames(template))
        template[] <- t(temp_array)
        template <- cbind(description = "", template) 
        template[1,1] <- desc
        names(dimnames(template)) <- dname
        return(template)
          # ------------------------------------------------
          # ------------------------------------------------
          # ------------------------------------------------
    }else{

        if (modality_1_index == 2) {  # dimensions are inverted
      temp_array <- t(temp_array) # R remplit colonne par colonne
      if (!is.na(array_values[1, 2])) {
        # Découper les valeurs par les virgules
        value_list <- as.numeric(unlist(strsplit(array_values[1, 2], ",")))
        # Remplir la matrice avec les valeurs extraites
        temp_array[] <- value_list  # Remplir la matrice avec les valeurs extraites
      }
    } else if (modality_1_index == 1) {
            if (!is.na(array_values[1, 2])) { 
        # Découper les valeurs par les virgules
        value_list <- as.numeric(unlist(strsplit(array_values[1, 2], ",")))
        # Remplir la matrice avec les valeurs extraites
        temp_array <- t(temp_array) # on a extrait les lignes, donc il faut inverser pour remplir correctement 
        temp_array[] <- value_list  # Remplir la matrice avec les valeurs extraites
        temp_array <- t(temp_array) # On remet à l'endroit, les dimensions n'étaient pas inversées.
      }
        }
        dname <- names(dimnames(template))
        template[] <- temp_array                        
        template <- cbind(description = "", template) 
        template[1,1] <- desc
        names(dimnames(template)) <- dname
        return(template)
    }
                        
                        } else {

    # -----------------------------------------
    # ---------- 1D, 2D, 3D CASE --------------  
    # -----------------------------------------
  
    
    # Find templates in the global environment
    template_names <- grep("^template_", ls(envir = .GlobalEnv), value = TRUE)
    templates <- lapply(template_names, function(nm) get(nm, envir = .GlobalEnv))
    names(templates) <- template_names
  
    # Si il y a une seule dimension, on cherche un vecteur
    if (length(dim_names) == 1) {

          dim_key <- tolower(dim_names) 
    
      # Rechercher les templates avec nom du type "template_<dim>_X"
      matching_names <- grep(paste0("^template_", dim_key, "_[^_]{1}$"), template_names, value = TRUE)

      if (length(matching_names) == 0) {
        stop("❌ No matching template found with dimension:", dim_key)
      }
      if (length(matching_names) > 1) {
        stop("❌ Multiple matching templates found for dimension: ", dim_key, ". Please ensure uniqueness.")
      }
      
      template <- templates[[matching_names[[1]]]]
          
        desc <- gsub("\\t", "", txt)
        # Extraire tout ce qui vient après le premier ~
        desc <- str_replace(desc, ".*?~", "")
        # Si desc est vide (ou contient juste des espaces), mettre NA
        if (str_trim(desc) == "") {
          desc <- NA
        }
                                
        txt <- gsub("\\s+", "", txt)
        txt <- unlist(strsplit(txt, "\\~"))
        txt <- txt[1]      
        
        # Traitement des données du fichier
      #data <- str_match(blocks, "\\[([^\\]]+)\\]=([^~]+)~([^|]+)\\|")
        data <- str_match(txt, "=(.+)")
                        
      values <- as.numeric(unlist(strsplit(data[1, 2], ",")))

        
      dname <- dim_names[1]
        desc_col <- rep("", length(template))
        temp_name <- names(template)
        template <- cbind("description"=desc_col,"value"=values) 
        template[1,1] <- desc
        rownames(template) <- temp_name
      return(template)
        
    } else {
      # -----------------------------------------
      # -----------------------------------------  

        desc <- unlist(strsplit(txt, "\\~\\|"))    
        desc <- gsub("\\t", "", last(desc))
        desc <- str_replace(desc, ".*?~", "")
        # Si desc est vide (ou contient juste des espaces), mettre NA
        if (str_trim(desc) == "") {
            desc <- NA
        }
        
                            
    txt <- gsub("\\s+", "", txt)
    blocks <- unlist(strsplit(txt, "\\~\\|"))
        
      # Si il y a plusieurs dimensions, on recherche le template 2D ou 3D
      matching_templates <- Filter(function(tmpl) {
        is.array(tmpl) &&
        length(dimnames(tmpl)) == length(dim_names) &&
        setequal(names(dimnames(tmpl)), dim_names)
      }, templates)
      
      # Erreur si aucun template n'est trouvé ou plusieurs templates trouvés
      if (length(matching_templates) == 0) {
        stop("❌ No matching template found with dimensions: ", paste(dim_names, collapse = ", "))
      }
      if (length(matching_templates) > 1) {
        stop("❌ Multiple matching templates found. Please ensure uniqueness.")
      }
      
      # Sélectionner le template correspondant
      template <- matching_templates[[1]]
      tmpl_dims <- dimnames(template)
      
      modality_map <- list()
      dimension_map <- list()

      for (block in blocks) {
        name_match <- str_match(block, "\\[([^\\]]+)\\]=([^~]+)~(.*)")
        if (is.na(name_match[1, 1])) next

        modals <- trimws(unlist(strsplit(name_match[1, 2], ",")))
        values <- as.numeric(unlist(strsplit(name_match[1, 3], ",")))
        ndim <- length(modals)

        positions <- integer(ndim)
        vector_dim <- NULL

        for (i in seq_along(modals)) {
          original_modal <- modals[i]

          # 1. Vérifier les mappings
          if (original_modal %in% names(modality_map)) {
            modals[i] <- modality_map[[original_modal]]
          } else if (original_modal %in% names(dimension_map)) {
            vector_dim <- dimension_map[[original_modal]]
            modals[i] <- NA
            positions[i] <- NA
            next
          }

          # 2. Est-ce une vraie modalité ?
          match_found <- FALSE
          for (j in seq_along(tmpl_dims)) {
            if (!is.na(modals[i]) && modals[i] %in% tmpl_dims[[j]]) {
              positions[i] <- j
              match_found <- TRUE
              break
            }
          }

          # 3. Si inconnue, interaction
          while (!match_found) {
            message("\n⚠️  Unknown modality detected: '", modals[i], "'")
            cat("Available options by dimension:\n")
            for (j in seq_along(tmpl_dims)) {
              cat(" - ", names(tmpl_dims)[j], ": ", paste(tmpl_dims[[j]], collapse = ", "), "\n")
            }

            response <- readline(paste0("❓ Is '", modals[i], "' a misspelled modality or a dimension name? (m/d): "))

            if (tolower(response) == "m") {
              new_modal <- readline(paste0("🔁 Replace modality '", modals[i], "' with: "))
              for (j in seq_along(tmpl_dims)) {
                if (new_modal %in% tmpl_dims[[j]]) {
                  modality_map[[original_modal]] <- new_modal
                  modals[i] <- new_modal
                  positions[i] <- j
                  match_found <- TRUE
                  break
                }
              }
              if (!match_found) {
                message("❌ '", new_modal, "' is not a valid modality. Try again.")
              }

            } else if (tolower(response) == "d") {
              new_dim <- readline(paste0("📐 Which dimension does '", modals[i], "' refer to? "))
              if (new_dim %in% names(tmpl_dims)) {
                dimension_map[[original_modal]] <- new_dim
                vector_dim <- new_dim
                modals[i] <- NA
                positions[i] <- NA
                match_found <- TRUE
                break
              } else {
                message("❌ '", new_dim, "' is not a valid dimension name.")
              }

            } else {
              message("❌ Invalid choice. Please type 'm' or 'd'.")
            }
          }
        }

        # --- Build index ---
        index <- vector("list", length(tmpl_dims))
        names(index) <- names(tmpl_dims)
        vector_dim <- unique(vector_dim)
        if (!is.null(vector_dim)) {
          # --- vector assignment over full dimension ---
          for (i in seq_along(modals)) {
            if (!is.na(modals[i])) {
              dim_index <- positions[i]
              index[[dim_index]] <- modals[i]
            }
          }
          index[[vector_dim]] <- tmpl_dims[[vector_dim]]

          if (length(values) != length(tmpl_dims[[vector_dim]])) {
            # Vérifier si la dimension a une seule valeur et recycler
            if (length(values) == 1) {
              recycled_value <- values
              values <- rep(values, length(tmpl_dims[[vector_dim]]))
              message("ℹ️ Recycled value ", recycled_value, " for dimension '", vector_dim, "' to match the full length of the dimension.")
            } else {
              warning("⚠️ Length mismatch: values=", length(values), " vs dimension '", vector_dim, "'=", length(tmpl_dims[[vector_dim]]))
            }
            next
          }

          template <- tryCatch({
            do.call("[<-", c(list(template), index, list(value = values)))
          }, error = function(e) {
            warning("⚠️ Error assigning block: ", block, " → ", e$message)
            template
          })

        } else {
          # --- standard case ---
          for (i in seq_along(modals)) {
            index[[positions[i]]] <- modals[i]
          }
          for (i in seq_along(index)) {
            if (is.null(index[[i]])) index[[i]] <- TRUE
          }

          template <- tryCatch({
            do.call("[<-", c(list(template), index, list(value = values)))
          }, error = function(e) {
            warning("⚠️ Error assigning block: ", block, " → ", e$message)
            template
          })
        }
      }
    n_dims <- length(dimnames(template))
    dname <- names(dimnames(template))
    if(n_dims == 2){
        template <- cbind(description = "", template) 
        template[1,1] <- desc
        names(dimnames(template)) <- dname
        return(template)
    }else if(n_dims ==3){
        attributes(template)$desc <- desc 
        return(template)
    }
        } # Fin du else 2D 3D 
                        }  # Fin du else 1D 2D 3D
                        }  # Fin de la fonction parse_vensim_data

# --------------------------------------------------------------------------------------------------

export_to_input_folder <- function(res, varname) {
  outputfolder <- here::here("notebooks", "r-nb", "inputs", "economy/")

  if (!dir.exists(outputfolder)) {
    dir.create(outputfolder, recursive = TRUE)
  }
    if("value" %in% colnames(res)){# 1D case
        export_name <- varname
        file_path <- file.path(outputfolder, paste0(export_name, ".csv"))
        first_dim_name <- dim_names[1]
        first_dim_mod <- rownames(res)
        res_export <- data.frame(first_dim_mod, res, check.names = FALSE, stringsAsFactors = FALSE)
        res_export <- res_export %>% select(description, first_dim_mod, everything())
        colnames(res_export)[2] <- first_dim_name

        res_export$value <- as.numeric(res_export$value)
                                
        write.csv(res_export, file_path, row.names = FALSE)
        cat("✅ Exported to input/:", export_name, "\n")
        print(res_export[,-1], row.names = FALSE)
        message("📂 See ", file_path, " for details.")
    }else{
        
      n_dim <- length(dimnames(res))
      first_dim_mod <- dimnames(res)[[1]]
      first_dim_name <- tolower(names(dimnames(res))[1])
    
      if (n_dim <= 2) {
        export_name <- varname
        file_path <- file.path(outputfolder, paste0(export_name, ".csv"))

        res_export <- data.frame(first_dim_mod, res, check.names = FALSE, stringsAsFactors = FALSE)
        res_export <- res_export %>% select(description, first_dim_mod, everything())
        colnames(res_export)[2] <- first_dim_name
          
        res_export[ , -(1:2)] <- lapply(res_export[ , -(1:2)], function(x) as.numeric(gsub('"', '', x)))

        write.csv(res_export, file_path, row.names = FALSE)
    
        cat("✅ Exported:", export_name, "\n")
        print(res_export[,-1], row.names = FALSE)
        message("📂 See ", file_path, " for details.")
      } else {
        last_dim <- names(dimnames(res))[n_dim]
        last_modalities <- dimnames(res)[[last_dim]]
    
        for (m in last_modalities) {
            export_name <- paste0(varname, "_", m)
          res2D <- res[, , m]
    
          assign(export_name, res2D, envir = .GlobalEnv)
          file_path <- file.path(outputfolder, paste0(export_name, ".csv"))

          res_export <- data.frame(first_dim_mod, res2D, check.names = FALSE, stringsAsFactors = FALSE)

          res_export <- res_export %>% mutate(description="")
            res_export$description[1] <- attr(res, "desc")

          res_export <- res_export %>% select(description, first_dim_mod, everything())
          colnames(res_export)[2] <- first_dim_name
        
          write.csv(res_export, file_path, row.names = FALSE)
    
          cat("✅ Exported to input/:", export_name, "\n")
          print(res_export[,-1], row.names = FALSE)
          message("📂 See ", file_path, " for details.")
        
        }
      }
    }
}
```
